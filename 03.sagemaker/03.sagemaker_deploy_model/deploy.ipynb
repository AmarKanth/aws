```python
from sagemaker.serve import ModelBuilder
from sagemaker.core.image_uris import retrieve


inference_image = retrieve(
    framework="sklearn",
    region="ap-south-1",
    version="1.2-1",
    image_scope="inference"
)


model_builder = ModelBuilder(
    model_path="s3://my-sagemaker-demo-bucket-amar/model-artifacts/model.tar.gz",
    image_uri=inference_image,
    role_arn="arn:aws:iam::058264205719:role/SageMakerDemoExecutionRole",
    env_vars={"SAGEMAKER_PROGRAM": "inference.py"}
)


model_builder.build()
predictor = model_builder.deploy(
    instance_type="ml.t3.medium",
    initial_instance_count=1
)
```
